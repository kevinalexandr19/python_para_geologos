### <span style="color:lightgreen">**¿Qué es una variable geólogica?**</span>
***
Una variable geológica es cualquier propiedad o característica que puede ser medida, observada o registrada en el contexto del estudio de la Tierra y los materiales geológicos.

Las variables geológicas pueden ser:
- <span style="color:gold">Cuantitativas</span>, cuando representan valores numéricos, o
- <span style="color:gold">Cualitativas</span>, cuando representan categorías o clases de información.

<span style="color:#43c6ac">Las variables geológicas pueden ser utilizadas para caracterizar o inferir información sobre los procesos y la historia geológica de una región o de una roca en particular. </span>

> Algunos ejemplos de variables geológicas incluyen:
>
> - Composición química y mineralógica de una roca o mineral.
> - Textura y estructura de una roca, incluyendo la orientación de sus componentes.
> - Propiedades físicas, como la densidad, la porosidad, la permeabilidad, la dureza o la resistencia.
> - Propiedades geofísicas, como la conductividad eléctrica, la velocidad de ondas sísmicas, la magnetización o la gravedad.
> - Edad, tanto relativa como absoluta, de los materiales geológicos.
> - Características de los depósitos sedimentarios, como la granulometría, la forma de los granos y la estratificación.
> - Patrones de deformación y fracturación en rocas y estructuras geológicas.

Las variables geológicas son esenciales para la descripción, la interpretación y la modelización de la Tierra y sus procesos, y son ampliamente utilizadas en exploración, minería, geofísica y otras disciplinas relacionadas con la Geología.

### <span style="color:lightgreen">**¿Cómo puedo representar variables en Python?**</span>
***

En Python, podemos manipular valores numéricos de la misma forma en la que usamos nuestra calculadora del celular.

Por ejemplo, podemos hacer sumas y restas:

> **Antes de ejecutar:** intenta calcular el resultado de cada operación. Luego, ejecuta el bloque con `Shift + Enter` y compara tu respuesta con la de Python.

```python
5 + 5
```

```python
10 - 4
```

Multiplicaciones y divisiones:

```python
3 * 3
```

```python
10 / 2
```

Potencias de un número:

```python
4 ** 2
```

```python
16 ** 0.5
```

O también, calcular el residuo de una división:
> El símbolo `%` se utiliza en Python para obtener el residuo de una división. En este ejemplo usamos números enteros, aunque también puede utilizarse con floats.

```python
10 % 3
```

El resultado de `10 % 3` es `1`: podemos formar tres grupos de `3` y queda una unidad. El símbolo `%` representa el residuo; no convierte el número a porcentaje.

Hasta ahora hemos escrito valores y operaciones directamente. Para usar un valor en una tarea posterior, podemos asociarlo a un nombre.
> En Python, el nombre de una variable es un **identificador**.

Crearemos una variable llamada `x` para almacenar el número 5:

> **Antes de ejecutar:** identifica el nombre, el signo de asignación y el valor. ¿Crees que esta línea mostrará un resultado?

```python
x = 5
```

Si escribimos la variable `x` en el siguiente bloque de código y luego lo ejecutamos (usando `Shift` + `Enter`), podremos observar el valor asociado a esta variable:

```python
x
```

**Revisemos la asignación** <br>
La instrucción `x = 5` asocia el nombre `x` al valor `5`. La asignación no muestra un resultado por sí sola; al ejecutar `x`, consultamos el valor asociado a ese nombre.

> El signo `=` se lee aquí como **asignación**. Python evalúa lo que está a la derecha y luego lo asigna al nombre de la izquierda. No es una igualdad matemática que se mantenga actualizada automáticamente.

En Python, las variables se definen asignándoles un valor utilizando el signo igual `=`.

El nombre de una variable puede ser cualquier cadena de caracteres alfanuméricos y guiones bajos, pero no puede comenzar con un número y no puede ser una **palabra reservada**.

>**¿Qué son las palabras reservadas?** <br>
> Las <span style="color:gold">palabras reservadas</span> de Python son aquellas que tienen un significado especial y están reservadas para un propósito específico y no se pueden utilizar para nombrar variables, funciones o cualquier otro objeto definido por el usuario.

<span style="color:#43c6ac">Es preferible que el nombre de una variable sea significativa y descriptiva de la tarea que realizamos.</span>

***
**Ejemplo: Muestreo de rocas** <br>
Si tengo 25 muestras de roca, podemos representar este valor en Python usando una variable llamada `muestras` y asignarle un valor de 25:

> Python distingue mayúsculas y minúsculas: `muestras` y `Muestras` son nombres diferentes. Un nombre descriptivo ayuda a leer el código, pero no define por sí solo el tipo de dato ni sus unidades.

```python
muestras = 25
```

Una vez creada esta variable, podemos usarla en una tarea posterior:

```python
muestras
```

```python
x, y, z = 1, 0, 1
```

```python
x
```

```python
y
```

```python
z
```

Los valores se asignan en el mismo orden: `x` recibe `1`, `y` recibe `0` y `z` recibe `1`.

### <span style="color:lightgreen">**Variables cuantitativas en Geología**</span>
***

En Python, los números enteros son conocidos como <span style="color:gold">integer</span>.

En un ejemplo anterior, usamos un integer para representar una cierta cantidad de muestras:

```python
25
```

<span style="color:#43c6ac">Si escribimos una línea de código debajo de otra y luego ejecutamos el bloque de código, Python empezará ejecutando la línea superior hasta llegar a la última línea en la parte inferior.</span>

Por ejemplo, crearemos dos variables `a` y `b` y los sumaremos en la última línea:

> **Antes de ejecutar:** sigue las instrucciones en orden e identifica qué valor tendrá cada variable cuando Python llegue a la suma.

```python
a = 3
b = 2
a + b
```

En un notebook, el resultado de la última expresión de una celda suele mostrarse automáticamente. En este caso, el resultado de `a + b` es `5`.

> **Revisemos el código:** primero se asigna `3` a `a`, luego `2` a `b` y finalmente se suman los valores asociados a ambos nombres. La suma no cambia por sí sola los valores de `a` y `b`.

***
**Pregunta: ¿Cómo representamos datos numéricos continuos como la concentración de Au en una muestra de roca o la longitud de una falla?** <br>
Respuesta: para poder usar cantidades con notación decimal, debemos hacer uso de otro tipo de dato numérico conocido como <span style="color:gold">float</span>.

> <span style="color:#43c6ac">La gran mayoría de valores cuantitativos en Geología (y también en el mundo real) poseen notación decimal.</span>

Por ejemplo, el valor de Pi redondeado a 4 decimales puede ser representado con un float:

```python
3.1416
```

***
**Ejemplo: Variables geológicas** <br>
Si la ley de un mineral es de 1.5 g/t, podemos representar este valor en Python usando una variable llamada `ley` y asignarle un valor de 1.5:

```python
ley = 1.5
ley
```

> Ten en cuenta que las unidades (g/t) no fueron representadas dentro del código.

Si la longitud de una falla es de 5.45 metros, podemos representar este valor en Python usando una variable llamada `longitud` y asignarle un valor de 5.45:
> Python no deduce las unidades a partir del número. Para interpretar `ley = 1.5`, debemos saber que en este ejemplo representa g/t; el mismo número podría representar otra propiedad o unidad.

```python
longitud = 5.45
longitud
```

Si las dimensiones de una veta de Au son de 40m x 50m x 10m, podemos representar el volumen de la veta en Python usando una variable llamada `volumen`:

> **Antes de ejecutar:** ¿qué operación representa el código y en qué unidades estará su resultado? Aquí estamos aproximando la veta como un bloque rectangular.

```python
volumen = 40 * 50 * 10
volumen
```

El volumen calculado es de **20 000 m³**, porque multiplicamos tres longitudes expresadas en metros.

> Una operación puede ejecutarse sin errores y aun así utilizar datos o unidades incorrectas. Por ejemplo, no deberíamos mezclar metros y centímetros en este cálculo sin convertirlos primero.

***
**Pregunta: ¿Podemos transformar integers a floats y viceversa?** <br>
Respuesta: Sí, podemos transformarlos a través de las funciones `int` y `float`:

```python
float(6)
```

Al transformar un float a integer, Python elimina la parte decimal y se aproxima hacia cero. No está redondeando al entero más cercano:

> **Antes de ejecutar:** ¿qué valor esperas obtener de `int(0.123)`?

```python
int(0.123)
```

El resultado es `0`. De la misma forma, `int(2.9)` produce `2` e `int(-2.9)` produce `-2`. Convertir a integer puede eliminar información de una medición; no significa que debamos hacerlo con todos los datos cuantitativos.

<span style="color:#43c6ac">En Python, los valores integer y float pueden combinarse:</span>

```python
3.14 + 5
```

***
**Ejemplo: Promedio aritmético** <br>
Supongamos que tenemos un conjunto de datos que representan la densidad de diferentes rocas en una cantera.\
Para trabajar con estos datos en Python, podemos crear variables que contengan cada uno de los valores:

> Para este ejemplo, todas las densidades están expresadas en **g/cm³**. Antes de calcular un promedio, debemos comprobar que los valores utilizan las mismas unidades.

```python
densidad_1 = 2.5
densidad_2 = 3.1
densidad_3 = 2.7
densidad_4 = 2.9
```

En este ejemplo, hemos creado cuatro variables diferentes, cada una de las cuales almacena un valor de densidad para una roca específica.\
Estas variables pueden utilizarse para realizar cálculos, por ejemplo, podemos calcular la densidad media de todas las rocas de la siguiente manera:

> **Antes de ejecutar:** ¿por qué la suma está entre paréntesis? ¿Qué ocurriría si solo dividiéramos la última densidad entre `4`?

```python
densidad_media = (densidad_1 + densidad_2 + densidad_3 + densidad_4) / 4
densidad_media
```

El promedio es aproximadamente **2.8 g/cm³**. Para llegar a este resultado, primero se suman las cuatro densidades y después se divide entre el número de muestras.

***
**Pregunta: ¿Qué pasa si corregimos una de las densidades?** <br>
Supongamos que la densidad de la primera muestra era de `3.3` g/cm³. Antes de ejecutar el siguiente bloque, ¿crees que `densidad_media` se actualizará por sí sola?

```python
# Corregimos la densidad de la primera muestra
densidad_1 = 3.3

# Consultamos el promedio calculado anteriormente
densidad_media
```

La variable `densidad_media` conserva el resultado calculado anteriormente. Cambiar `densidad_1` no vuelve a ejecutar la fórmula.

| Momento | `densidad_1` | `densidad_media` |
| --- | --- | --- |
| Antes de la corrección | 2.5 | Aproximadamente 2.8 |
| Después de cambiar la densidad | 3.3 | Aproximadamente 2.8 |

Para obtener el promedio con el dato corregido, debemos repetir el cálculo:

```python
# Calculamos nuevamente el promedio con la densidad corregida
densidad_media = (densidad_1 + densidad_2 + densidad_3 + densidad_4) / 4
densidad_media
```

Ahora el resultado es **3.0 g/cm³**.

> **Revisemos el orden:** cambiar un dato y recalcular un resultado son dos instrucciones diferentes. Si vuelves a ejecutar la celda donde definimos las cuatro densidades originales, `densidad_1` recibirá nuevamente el valor `2.5`.

### <span style="color:lightgreen">**Variables cualitativas en Geología**</span>
***

¿Qué pasa si tengo el nombre de una roca?, ¿qué hay de su textura?, ¿o del tipo de alteración que presenta?, ¿cómo podemos representar estos datos categóricos dentro de Python?

Para hacer esto, necesitamos emplear un tipo de dato llamado <span style="color:gold">string</span>:

```python
nombre = "andesita"
nombre
```

Un string se caracteriza por estar encasillado entre comillas dobles `""` o simples `''`.

> <span style="color:#43c6ac">Podemos colocar todo tipo de caracteres dentro de un string.</span>

Por ejemplo, podemos insertar una frase dentro de un string:

> **Revisemos el código anterior:** `nombre` es el identificador y `"andesita"` es el valor de tipo string. Las comillas forman parte de la escritura del texto en Python; no se colocan alrededor del nombre de la variable.

```python
frase = "El presente es la clave del pasado"
frase
```

O también podemos usar comillas triples para insertar un párrafo compuesto por varias líneas de texto:

```python
"""
Una roca es un agregado de minerales
El presente es la clave del pasado
La geología estudia la Tierra y sus fenómenos
"""
```

> Si vemos la representación del texto en Python, notamos un símbolo `\n` que representa una continuación en la siguiente línea del párrafo.

***
**Ejemplo: Variable geológica cualitativa** <br>
Si la alteración de una roca es clasificada como propilítica, podemos representar este valor en Python usando una variable llamada `alteracion` y asignarle el valor "propilítica":

> Es recomendable no usar tildes `´` en los nombres de las variables en Python.
> Esta recomendación se refiere a los identificadores. El contenido de un string sí puede incluir tildes, como ocurre en `"propilítica"`.

```python
alteracion = "propilítica"
alteracion
```

<span style="color:#43c6ac">Otra propiedad de los strings es que se pueden sumar:</span>

En este caso, el símbolo `+` une fragmentos de texto. Esta operación se conoce como **concatenación**.

> **Antes de ejecutar:** observa los strings `" "`. ¿Qué función cumplen los espacios en la frase final?

```python
"La andesita" + " " + "es una roca" + " " + "de textura afanítica"
```

***
**Pregunta: ¿Se pueden transformar números en strings y viceversa?** <br>
Respuesta: Sí, podemos usar las funciones `int`, `float` y `str`, según el tipo de dato que necesitamos obtener.

> Para convertir un string, su contenido debe tener un formato numérico válido para la función utilizada. Por ejemplo, `float("1.5")` admite un punto decimal, mientras que `int("10")` representa un entero. También pueden aparecer signos, como en `float("-1.5")`.

Antes de ejecutar los siguientes bloques, identifica qué tipo de dato recibiremos como resultado.

```python
float("5")
```

```python
int("10")
```

**Pregunta: ¿Podemos convertir `"1.5 g/t"` directamente con `float`?** <br>
Respuesta: No, porque el string también contiene las unidades. De la misma forma, `int("1.5")` no interpreta directamente ese texto como un entero. Ambas conversiones producirían un `ValueError`.

> Identificar el formato del dato es parte de comprender el código. La conversión necesita un valor válido para la función que estamos utilizando.

Para pasar de números a strings usaremos la función `str`:

> **Revisemos la diferencia:** `5 + 5` produce `10`, pero `"5" + "5"` produce `"55"`. Aunque veamos los mismos dígitos, el tipo de dato cambia el significado de la operación.

```python
str(5.0)
```

```python
str(100)
```
