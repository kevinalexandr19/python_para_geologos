# **Instalar Python y JupyterLab**

Vamos a preparar tu computadora para abrir notebooks y comprender cómo funciona su código.

**Versión de referencia: Python 3.15.2**, última versión estable verificada el **15 de noviembre de 2026** en [Python.org](https://www.python.org/downloads/release/python-3152/). Antes de descargar, revisa [Downloads](https://www.python.org/downloads/): si hay una versión estable más reciente, usa esa versión. Evita versiones con etiquetas **alpha**, **beta** o **rc**.

Usaremos **Python oficial + venv + pip**. La ruta principal es para **Windows 10/11**; también encontrarás indicaciones para macOS y Linux.

> **¿Estás usando JupyterLite?** Para estudiar en el navegador no necesitas instalar Python. Esta guía es opcional y sirve para trabajar en tu computadora. El entorno local y el kernel de JupyterLite son distintos.

## **Antes de empezar**

| Herramienta | ¿Para qué sirve? |
| --- | --- |
| **Python** | Ejecuta las instrucciones del código. |
| **venv** | Crea un entorno separado para tu proyecto. Viene con Python. |
| **pip** | Instala librerías dentro del entorno seleccionado. |
| **JupyterLab** | Permite abrir notebooks y ejecutar sus celdas desde el navegador. |
| **Kernel** | Es el proceso de Python que ejecuta las celdas. |



Los comandos de instalación se escriben en la **terminal**, una línea a la vez. Las instrucciones de Python se escriben en las **celdas de código** del notebook. No copies símbolos como `PS C:\...>` o `>>>`.

## **Paso 1. Descargar e instalar Python**

1. Abre [la página oficial de descargas](https://www.python.org/downloads/).
2. Comprueba la versión estable y abre su página de lanzamiento.
3. En **Files**, selecciona **Windows installer (64-bit)** para un equipo habitual con procesador Intel o AMD. Para Windows sobre ARM, elige **ARM64**. Puedes consultar el tipo de sistema en **Configuración → Sistema → Acerca de**.
4. Ejecuta el instalador. Activa **Add python.exe to PATH** si aparece y conserva la instalación de **pip** y del launcher.
5. Elige **Install Now** y conserva la ubicación sugerida para tu usuario; no necesitas crear `C:\Python`.

La página también ofrece **Python install manager**, una herramienta oficial para instalar y actualizar Python en Windows. Si elegiste ese botón, instala el administrador, abre una terminal nueva y ejecuta:

```powershell
py install 3.15
py -V:3.15 --version
```

Esto instala la revisión disponible de la serie 3.15. Si la serie estable cambió, sustituye `3.15` por la que acabas de verificar.

**Comprobación:** abre una terminal nueva buscando **PowerShell** en Inicio:

```powershell
py -3.15 --version
```

Con la versión de referencia deberías ver `Python 3.15.2`. Adapta el selector `-3.15` si elegiste otra serie. El selector permite elegir qué Python usar cuando tienes varias versiones.

> Si aparece `>>>`, entraste al intérprete interactivo. Escribe `exit()` y presiona Enter para volver a la terminal.

## **Paso 2. Crear la carpeta del proyecto y su entorno**

Abre la carpeta donde quieras guardar tus proyectos en el Explorador. Escribe `powershell` en su barra de dirección y presiona Enter. Después ejecuta:

```powershell
mkdir geologia_python
cd geologia_python
py -3.15 -m venv .venv
```

**¿Qué hace cada línea?** `mkdir` crea una carpeta, `cd` entra en ella y `-m venv .venv` pide al Python seleccionado que cree un entorno llamado `.venv`.

Tu carpeta se verá así cuando agregues tus materiales:

```text
geologia_python/
├── .venv/                 ← Python y librerías del entorno
├── mi_primer_notebook.ipynb
└── datos/                 ← tus CSV y otros archivos
```

Guarda los notebooks y datos **fuera de `.venv`**. Usaremos la ruta explícita del Python del entorno; no necesitas activar scripts ni cambiar la política de ejecución de PowerShell.

```powershell
.\.venv\Scripts\python.exe --version
.\.venv\Scripts\python.exe -m pip --version
```

**Comprobación:** la primera línea muestra la versión elegida. La segunda muestra pip y una ruta que contiene `geologia_python\.venv`.

## **Paso 3. Instalar JupyterLab y las primeras librerías**

Desde la misma carpeta, ejecuta una línea a la vez y espera a que termine:

```powershell
.\.venv\Scripts\python.exe -m pip install --upgrade pip
.\.venv\Scripts\python.exe -m pip install jupyterlab ipykernel numpy pandas matplotlib
.\.venv\Scripts\python.exe -m pip check
```

**JupyterLab** abre notebooks; **ipykernel** permite que el entorno ejecute sus celdas. **NumPy**, **pandas** y **Matplotlib** permiten trabajar con arreglos, tablas y gráficos.

En `python.exe -m pip install pandas`, indicamos **qué Python** debe ejecutar pip. Así evitamos instalar una librería en un entorno y buscarla desde otro.

**Comprobación:** `pip check` debería mostrar `No broken requirements found.` Esto comprueba dependencias declaradas; en el paso 5 también comprobaremos las importaciones.

> Este conjunto inicial no incluye todas las librerías especializadas del repositorio. Si una librería todavía no soporta la versión estable recién publicada de Python, consulta su documentación. No necesitas instalar todas las librerías para comenzar.

## **Paso 4. Abrir JupyterLab**

En la misma terminal:

```powershell
.\.venv\Scripts\python.exe -m jupyterlab
```

1. Espera a que se abra el navegador. Si no ocurre, copia la dirección local que muestra la terminal, normalmente `http://localhost:8888/lab?...`.
2. En el **Launcher**, busca **Notebook** y selecciona **Python 3 (ipykernel)**.
3. Cambia el nombre a `mi_primer_notebook.ipynb` y guarda con **Ctrl + S**.

Mantén la terminal abierta mientras trabajas. Aunque veas JupyterLab en el navegador, este Python se ejecuta en tu computadora.

**Para abrir el curso:** descarga el repositorio con **Code → Download ZIP**, extrae el ZIP y copia su carpeta `notebooks` dentro de `geologia_python`. En JupyterLab, abre `notebooks/pg000_inicio.ipynb`. Conserva las carpetas `files` y `resources` junto a esos notebooks: contienen los datos e imágenes de las lecciones.

## **Paso 5. Verificar qué Python ejecuta las celdas**

Ejecuta las siguientes celdas con **Shift + Enter**. Antes de hacerlo, intenta explicar qué mostrará cada línea.

````python
import sys

print("Versión de Python:", sys.version.split()[0])
print("Ejecutable del kernel:", sys.executable)
````

**¿Qué debes observar?** En JupyterLab local, la ruta del ejecutable debería apuntar al entorno `.venv` creado en el paso 2. En JupyterLite verás una ruta del navegador: instalar paquetes en tu computadora no modifica ese kernel.

Ahora comprobaremos las librerías con una pequeña tabla geológica.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

muestras = pd.DataFrame({
    "Profundidad_m": [10, 20, 30],
    "Densidad_g_cm3": [2.6, 2.7, 2.8],
})

print(muestras)
print("Densidad media:", np.mean(muestras["Densidad_g_cm3"]))

ax = muestras.plot(x="Densidad_g_cm3", y="Profundidad_m", marker="o")
ax.invert_yaxis()
ax.set_xlabel("Densidad (g/cm³)")
ax.set_ylabel("Profundidad (m)")
plt.show()
````

**Predice y explica:** ¿qué cambia si la segunda profundidad pasa de 20 a 25 metros? ¿Qué columna se utiliza en cada eje? Cambia solo ese valor y ejecuta de nuevo la celda para comparar.

Si las importaciones y el gráfico funcionan en tu entorno local, ya puedes comenzar.

## **Problemas frecuentes**

| Lo que ves | Qué revisar |
| --- | --- |
| `py` no se reconoce | Abre otra terminal y revisa la instalación del launcher/administrador. Si `python --version` muestra la versión elegida, puedes crear el entorno con `python -m venv .venv`. |
| Versión inesperada | Comprueba `py -3.15 --version` antes de crear el entorno. Un entorno existente no cambia automáticamente al instalar otro Python. |
| No se encuentra `.venv\Scripts\python.exe` | Debes estar dentro de `geologia_python` y haber completado el paso 2. |
| `No module named pip` | En Windows local, prueba `.\.venv\Scripts\python.exe -m ensurepip --upgrade`. En Linux puede ser necesario instalar el soporte de la distribución. |
| `ModuleNotFoundError` al importar | Compara `sys.executable` con la ruta del entorno, instala usando ese ejecutable y reinicia el kernel. |
| pip informa incompatibilidad o intenta compilar | Comprueba el soporte de la librería para tu versión y sistema. Conserva el mensaje completo para consultar su documentación. |
| El navegador no abre JupyterLab | Usa la URL local que aparece en la terminal y comprueba que el proceso siga activo. |
| `SyntaxError` al instalar | El comando posiblemente se pegó en una celda o después de `>>>`; escríbelo en la terminal. |
