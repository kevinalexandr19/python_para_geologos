### <span style="color:lightgreen">**¿Qué significa bootstrap?**
***

En estadística, el <span style="color:gold">bootstrap</span> o <span style="color:gold">remuestreo</span> es un método robusto que permite realizar inferencias sobre una población a partir de muestras pequeñas o de distribución desconocida. 

Si tenemos un conjuntos de datos, podemos realizar **bootstrap** para obtener un nuevo conjunto de datos que será extraído de manera aleatoria y con reemplazo a partir del conjunto original.





En la figura, cada número identifica una observación de la muestra original. Un mismo número puede aparecer varias veces en un remuestreo. La línea dorada marca la media original; el histograma reúne una media por remuestreo.



<br>
Este procedimiento estadístico permite simular las condiciones de muestreo original y calcular medidas como el error estándar, intervalos de confianza y pruebas de hipótesis.

Es importante tener en cuenta que este método no incrementa la cantidad de información en el conjunto de datos, solamente nos ayuda a <span style="color:lightgreen">obtener una mejor estimación acerca de la distribución de las estadísticas (media, varianza, etc.) de las muestras</span>.
***

El bootstrap habitual supone observaciones representativas e independientes. Con diez datos la cobertura de los intervalos puede ser limitada; sesgos de muestreo, dependencia espacial y valores extremos requieren revisión.

<a id="parte-2"></a>

### <span style="color:lightgreen">**Generación de datos**
***
En este ejemplo, crearemos un conjunto pequeño de datos y aplicaremos bootstrap para estimar la distribución de la media y varianza de dicho conjunto.

```python
import scipy
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Asumiremos que el conjunto tiene una distribución normal o gaussiana $N[0,1]$:

```python
# Número de muestras en el conjunto original
n = 10
# Número de conjuntos a crear con bootstrap
m = 1000

# Generador aleatorio de datos
rng = np.random.default_rng(42)

# Conjunto original vacío (lleno de ceros)
data = np.zeros(n)

# Bucle que reemplaza el valor de cada celda (cero) con un valor aleatorio (rng)
for i in range(0, n):
    data[i] = rng.normal() 
```

Observamos el conjunto original de 10 datos:

```python
# Conjunto de datos
print(data)
```

A partir de este conjunto original, tomaremos `n` muestras con reemplazo unas `m` veces:

Cada fila conserva n=10 observaciones. Muestrear con reemplazo permite repetir una observación y omitir otras; no genera valores nuevos fuera de la muestra.

```python
# Conjunto de bootstrap vacío (lleno de ceros) de tamaño m x n
bootstrap = np.zeros((m, n)) 

# Bucle que reemplaza el valor de cada celda (cero) con una valor aleatorio del conjunto original
for i in range(0, m):
    for sample in range(0, n):
        bootstrap[i, sample] = rng.choice(data)

print(f"El conjunto de remuestreo tiene un tamaño de {bootstrap.shape}")
```

En total, tenemos 1000 conjuntos de datos, cada uno de estos conjuntos contiene 10 muestras tomadas del conjunto original (con repetición).

Observamos los primeros 5 conjuntos generados:

```python
for array in bootstrap[:5]:
    print(array)
    print("")
```

### <span style="color:gold">**De una muestra a muchos remuestreos**</span>

El color identifica el valor original seleccionado, no una nueva medición. Cada fila tiene diez posiciones; repeticiones y ausencias son parte del muestreo con reemplazo.

```python
fig, ejes = plt.subplots(2, 1, figsize=(9,4.5), gridspec_kw={'height_ratios':[1,3]})
imagen_original = ejes[0].imshow(data[None,:], aspect='auto', cmap='viridis', vmin=data.min(), vmax=data.max())
ejes[0].set(yticks=[0], yticklabels=['Original'], xticks=np.arange(n), xlabel='Posición en la muestra')
ejes[1].imshow(bootstrap[:5], aspect='auto', cmap='viridis', vmin=data.min(), vmax=data.max())
ejes[1].set(yticks=np.arange(5), yticklabels=[f'Remuestreo {i+1}' for i in range(5)],
            xticks=np.arange(n), xlabel='Posición en cada remuestreo')
fig.colorbar(imagen_original, ax=list(ejes), label='Valor simulado (sin unidades)', fraction=.03, pad=.03)
fig.suptitle('Los remuestreos reutilizan las mismas observaciones')
fig.subplots_adjust(left=.18, right=.85, hspace=.7, top=.85)
plt.show()
```

<a id="parte-3"></a>

### <span style="color:lightgreen">**Cálculo de medidas estadísticas**
***
Ahora que tenemos los 1000 conjuntos de remuestreo, calcularemos los valores de media y varianza para cada uno de estos conjuntos:

```python
# Conjuntos vacíos (llenos de ceros) en donde almacenaremos la media y varianza de cada remuestreo
media = np.zeros(m)
varianza = np.zeros(m)

# Bucle que reemplaza el valor de cada remuestreo (cero) con su respectiva media y varianza
for i in range(0, m):
    media[i] = np.mean(bootstrap[i, :])
    varianza[i] = np.var(bootstrap[i, :], ddof=1)
```

Observamos parte de los resultados:

```python
# Media de los 10 primeros conjuntos
for mean in media[:10]:
    print(mean)
```

```python
# Varianza de los 10 primeros conjuntos
for var in varianza[:10]:
    print(var)
```

<a id="parte-4"></a>

### <span style="color:lightgreen">**Visualización de la incertidumbre**
***
Ahora que tenemos la media y varianza de los conjuntos generados a través de bootstrap, crearemos histogramas para visualizar su distribución:

La dispersión de las medias remuestreadas estima el error estándar de la media, no la dispersión de las observaciones individuales. La varianza de cada remuestreo se calculará con ddof=1.

```python
# Figura principal
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5), sharey=True)

# Histograma de la media
ax1.hist(media, bins=25, edgecolor="black")
ax1.set_title("Media - Bootstrap")
ax1.set_xlabel("Valor")
ax1.set_ylabel("Frecuencia")
ax1.grid()
ax1.set_axisbelow(True)
ax1.axvline(data.mean(), c="red") # Media del conjunto original

# Histograma de la varianza
ax2.hist(varianza, bins=25, edgecolor="black")
ax2.set_title("Varianza - Bootstrap")
ax2.set_xlabel("Valor")
ax2.grid()
ax2.set_axisbelow(True)
ax2.axvline(data.var(ddof=1), c="red") # Varianza del conjunto original

plt.tight_layout()
```

Por último, generaremos un resumen estadístico de los valores de media y varianza:

```python
df = pd.DataFrame({"Media": media, "Varianza": varianza})
df.describe()
```

### <span style="color:gold">**Error estándar e intervalo bootstrap**</span>

La desviación de las medias estima el error estándar. Los percentiles 2.5 y 97.5 dan un intervalo bootstrap percentil aproximado. No representan el intervalo que contiene el 95% de las observaciones originales.

```python
error_estandar = media.std(ddof=1)
ic_bootstrap = np.quantile(media, [.025,.975])
fig, ejes = plt.subplots(1,2,figsize=(10,3.8))
ejes[0].hist(data,bins=6,edgecolor='white',color='#0072B2')
ejes[0].set(title=f'Datos originales: n={n}',xlabel='Valor observado',ylabel='Número de observaciones')
ejes[1].hist(media,bins=25,edgecolor='white',color='#0072B2')
for limite in ic_bootstrap:ejes[1].axvline(limite,color='#D55E00',ls='--')
ejes[1].axvline(data.mean(),color='#009E73',label='Media observada')
ejes[1].set(title=f'Medias bootstrap: m={m}; SE≈{error_estandar:.3f}',
            xlabel='Media de un remuestreo',ylabel='Número de remuestreos')
ejes[1].legend()
fig.tight_layout()
plt.show()
print(f'IC percentil aproximado del 95% para la media: {ic_bootstrap}')
```

En repeticiones del proceso de muestreo, un procedimiento de confianza bien calibrado pretende cubrir el parámetro con la frecuencia indicada. Este intervalo concreto no asigna una probabilidad del 95% a un parámetro fijo. [Intervalos para la media: NIST](https://itl.nist.gov/div898/handbook/eda/section3/eda352.htm).

### <span style="color:lightgreen">**En conclusión...**
***
    
- Aumentar n (observaciones independientes) puede reducir el error estándar. Aumentar m (remuestreos) estabiliza su estimación, pero no agrega información a la muestra original.

- Usando bootstrap, el valor de incertidumbre para la media se aproxima al error estándar.

- Este método nos permite encontrar valores de incertidumbre para muchas medidas estadísticas; la validez depende del estimador, la muestra y los supuestos.

- El método de bootstrap no toma en cuenta el contexto espacial (e.g. ubicación de los datos, correlación espacial, etc.).
